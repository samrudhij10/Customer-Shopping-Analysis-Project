```python
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x :x.fillna(x.median()))
df['Review Rating']
```

```python
df.isnull().sum()
```

```python
df.columns= df.columns.str.lower()
df.columns= df.columns.str.replace(' ','_')
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
df.columns
```

```python
#create a column age wise .
df['age'].value_counts
labels=['young_adult','adult','middle_age','senior']
df['age_group']=pd.qcut(df['age'],q=4,labels=labels)  #(df['age'],q=4, labels=labels)
df[['age','age_group']]
```

```python
df['frequency_of_purchases'].value_counts
```

```python
# create columns purchase frequency days : 
days={
      'Fortnightly':14,
      'Weekly':7,
      'Monthly':30,
      'Quarterly':90,
      'Bi-weekly':14,
      'Annually':365,
      'Every 3 months':90
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].str.lower()
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(days)

df[['frequency_of_purchases','purchase_frequency_days']]
```
